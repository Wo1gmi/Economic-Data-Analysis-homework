Запустите ячейки по порядку. Ноутбук установит недостающие пакеты, скачает Parquet в рабочую папку и загрузит вакансии в объект `vacancies`.

```python
from pathlib import Path
from urllib.request import urlretrieve
import importlib.util
import subprocess
import sys

missing_packages = [
    package for package in ("pandas", "pyarrow")
    if importlib.util.find_spec(package) is None
]
if missing_packages:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing_packages])

import pandas as pd

data_url = "https://raw.githubusercontent.com/pparshakov/data/main/hh_vacancies_2023_2026.parquet"
data_file = Path("vacancies.parquet")

if not data_file.exists():
    urlretrieve(data_url, data_file)

vacancies = pd.read_parquet(data_file)
vacancies.head()
```

```python
vacancies_by_year = (
    vacancies["published_month"]
    .astype("string")
    .str.slice(0, 4)
    .value_counts()
    .sort_index()
    .rename_axis("year")
    .reset_index(name="vacancies")
)
vacancies_by_year
```

Данные загружены в объект `vacancies`. Таблица выше нужна только для проверки, что файл открылся. Дальше анализируйте данные своим способом.
